# Online Video Depth Anything

In the following we show a demo inference script.

```python
import sys
import os
import yaml
import torch

# Projekt-Root (one layer over notebooks/)
project_root = os.path.abspath(os.path.join(os.pardir))
if project_root not in sys.path:
    sys.path.insert(0, project_root)

from src.models.video_depth import onlineVideoDepthAnything
import cv2
import numpy as np 
from src.utils.loading_utils import load_video_as_numpy
from notebooks_utils import show_video_jshtml, show_depth_video_jshtml

```

```python
input_video = load_video_as_numpy(os.path.join(project_root, 'assets/example_videos/Cars_and_Gasstation.mp4'))
show_video_jshtml(input_video)
```

```python
model_config = os.path.join(project_root, 'configs/oVDA_c16.yaml')
with open(model_config, 'r') as f:
    config = yaml.safe_load(f)

oVDA = onlineVideoDepthAnything(**config['net'])

ckpt = torch.load(os.path.join(project_root, config['pretrained_path']), map_location='cpu')

oVDA.load_state_dict(ckpt)
oVDA.eval()
print()
```

```python

prediction = oVDA.infer_video_depth(input_video, device='cuda:4', preprocess_device='cpu', input_size=720, print_process_res=True, fp32=True, output_raw=False)
```

```python
show_depth_video_jshtml(prediction[0])
```
